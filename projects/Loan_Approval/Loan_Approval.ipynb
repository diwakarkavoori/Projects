# Univariate Analysis

```python
plotnumber=1
plt.figure(figsize=(16,8))
for i in df_cat:
    plt.subplot(3,3,plotnumber)
    sns.countplot(x=df_cat[i])
    plt.xticks(rotation=45)
    plt.tight_layout()
    plotnumber=plotnumber+1
```

## Insights

+ Most of the applicants are male.
+ Most applicants have credit history.
+ Most applicants are  graduates.
+ Most applicants are not self employed. 
+ A loan term of 360 seems to be the most common  choice for most applicants.
+ There seem to be more number of  applicants who are married.  
+ There seem to be more applicants from semi-urban  property area compared to urban and rural property areas. 
+ The number of applicants decreases as the number of dependents for the loan applicant increases. 

```python
plotnumber=1
plt.figure(figsize=(16,8))
for i in df_count:
    plt.subplot(1,3,plotnumber)
    sns.histplot(x=df_count[i],kde=True)
    plt.xticks(rotation=45)
    plt.tight_layout()
    plotnumber=plotnumber+1
```

# Bivariate Analyis

```python
plotnumber=1
plt.figure(figsize=(16,8))
for i in df_cat:
    plt.subplot(3,3,plotnumber)
    sns.countplot(x=df_cat[i],hue=df.Loan_Status)
    plt.xticks(rotation=45)
    plt.tight_layout()
    plotnumber=plotnumber+1
```

## Insights

+ More people who are male,have credit history and are graduates seem to get their loan approved.
+ Applicants who are not self employed seemed to be more successful in getting their loans approved. 
+ A loan term of 360 seems to be the most optimal choice for granting loans.
+ More people who are married seem to be getting their loan approved.
+ Semi Urban people seem to be more successful in getting their loan approved, compared to rural and urban people.
+ Applicants with 0 dependents seem to be most successful in getting their loans approved.
+ Applicants with 2 dependents seem to be more successful in getting their loans approved, compared to applicants with 1 dependent.
